# 06. EDA Summary & Modeling Logic

## 1. Data Reliability Confirmation

From preprocessing and SQL sanity checks:
- No missing values in critical features
- All physicochemical variables fall within realistic ranges
- No corrupted or invalid records detected

**Conclusion:** The dataset is reliable for downstream modeling.

## 2. Target Variable Characteristics (Wine Quality)

- Ordinal target variable (quality scores)
- Strong class imbalance (majority around 5–6)
- Overlapping feature distributions across classes

**Conclusion:
- Accuracy is misleading under class imbalance because it ignores minority classes.
- We choose F1-macro because every class is treated equally, regardless of frequency.

## 3. Feature Distribution Insights (Univariate)

- Most features are non-Gaussian and skewed
- Mild to moderate outliers present
- Distribution shifts observed across wine types.


**Decision:** 
- Outliers cannot removed because extreme values are considerd as special wines and as well as spoiled wines based on quality.
- Tree-based models preferred due to robustness to skewness and outliers.

## 4. Feature Relationships & Interactions (Bivariate)

- Predominantly non-linear relationships
- Strong interaction effects (e.g., alcohol–density, sugar–density)
- Individual features insufficient alone
- Heavy overlapping accors quality classes

**Decision:** Models must capture interactions without manual feature engineering.

## 5. Correlation & Multicollinearity Conclusions

- Most linear correlations are weak
- Some feature redundancy exists (e.g., sulfur dioxide features)
- High VIF values impact linear models

**Decision:** Retain correlated features and avoid linear models.Tress models are better to use.

## 6. Final Decisions

### Selected Models
- Random Forest
- Extra Trees
- XGBoost
- LightGBM

**Reason:** Ability to handle non-linearity, interactions, skewed distributions, and multicollinearity.

### Metric Selection
- Primary Metric: F1-macro

**Reason:** Treats all quality classes equally and Robust to class imbalance.

- Secondary Metric: RMSE

**Reason:** Penalizes large ordinal misclassifications
 
### Hyperparameter Optimization & Ensembling

- Hyperparameter tuning using Optuna
- Nested cross-validation to avoid optimistic bias
- Stacking ensemble to combine model strengths
