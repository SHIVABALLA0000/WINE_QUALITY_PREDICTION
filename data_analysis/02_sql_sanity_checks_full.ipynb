# 02. SQL Sanity Checks & Data Validation 
Goal:
- Validate **every feature** using MIN / MAX / AVG
- Detect impossible or suspicious values
- Build **business trust** in data before modeling

## 1. Database Connection

```python
import pandas as pd
from sqlalchemy import create_engine
import sqlite3
engine= sqlite3.connect("wine_quality.db")
```

## 2. Row Count Check

```python
pd.read_sql('SELECT COUNT(*) AS total_rows FROM wine_cleaned;', engine)
```

### Observation
- Confirms successful ingestion of both red and white wine datasets
- Acts as a baseline for all further analysis

## 3. Quality Distribution

```python
pd.read_sql('''SELECT quality, COUNT(*) AS count FROM wine_cleaned GROUP BY quality ORDER BY quality;''', engine)
```

### Observation
- Reveals class imbalance
- Middle quality scores dominate the dataset

## 4. Full Feature Range Validation

```python
pd.read_sql('''SELECT MIN(fixed_acidity) AS min_fixed_acidity, MAX(fixed_acidity) AS max_fixed_acidity, AVG(fixed_acidity) AS avg_fixed_acidity,
MIN(volatile_acidity) AS min_volatile_acidity, MAX(volatile_acidity) AS max_volatile_acidity, AVG(volatile_acidity) AS avg_volatile_acidity,
MIN(citric_acid) AS min_citric_acid, MAX(citric_acid) AS max_citric_acid, AVG(citric_acid) AS avg_citric_acid
FROM wine_cleaned;''', engine)
```

```python
pd.read_sql(""" SELECT MIN(chlorides) AS min_chlorides, MAX(chlorides) AS max_chlorides, AVG(chlorides) AS avg_chlorides,
MIN(free_sulfur_dioxide) AS min_free_sulfur_dioxide, MAX(free_sulfur_dioxide) AS max_free_sulfur_dioxide, AVG(free_sulfur_dioxide) AS avg_free_sulfur_dioxide,
MIN(total_sulfur_dioxide) AS min_total_sulfur_dioxide, MAX(total_sulfur_dioxide) AS max_total_sulfur_dioxide, AVG(total_sulfur_dioxide) AS avg_total_sulfur_dioxide,
MIN(sulphates) AS min_sulphates, MAX(sulphates) AS max_sulphates, AVG(sulphates) AS avg_sulphates
FROM wine_cleaned""",
engine)
```

```python
pd.read_sql(""" SELECT MIN(residual_sugar) AS min_residual_sugar, MAX(residual_sugar) AS max_residual_sugar, AVG(residual_sugar) AS avg_residual_sugar,
MIN(density) AS min_density, MAX(density) AS max_density, AVG(density) AS avg_density,
MIN(pH) AS min_pH, MAX(pH) AS max_pH, AVG(pH) AS avg_pH,
MIN(alcohol) AS min_alcohol, MAX(alcohol) AS max_alcohol, AVG(alcohol) AS avg_alcohol
FROM wine_cleaned;""", engine)
```

### Observation
- All acidity-related features fall within chemically valid ranges
- Sulphur_Dioxide,sulphates,chlorides having ranges which are general in wine making industries.
- Ph,alcohol,density,,residual_sugars are chemically valid ranges for a wine.

## 5. NULL Checks

```python
pd.read_sql('''SELECT
SUM(CASE WHEN fixed_acidity IS NULL THEN 1 ELSE 0 END) AS fixed_acidity_nulls,
SUM(CASE WHEN volatile_acidity IS NULL THEN 1 ELSE 0 END) AS volatile_acidity_nulls,
SUM(CASE WHEN citric_acid IS NULL THEN 1 ELSE 0 END) AS citric_acid_nulls,
SUM(CASE WHEN residual_sugar IS NULL THEN 1 ELSE 0 END) AS residual_sugar_nulls,
SUM(CASE WHEN chlorides IS NULL THEN 1 ELSE 0 END) AS chlorides_nulls,
SUM(CASE WHEN free_sulfur_dioxide IS NULL THEN 1 ELSE 0 END) AS free_sulfur_dioxide_nulls,
SUM(CASE WHEN total_sulfur_dioxide IS NULL THEN 1 ELSE 0 END) AS total_sulfur_dioxide_nulls,
SUM(CASE WHEN density IS NULL THEN 1 ELSE 0 END) AS density_nulls,
SUM(CASE WHEN pH IS NULL THEN 1 ELSE 0 END) AS pH_nulls,
SUM(CASE WHEN sulphates IS NULL THEN 1 ELSE 0 END) AS sulphates_nulls,
SUM(CASE WHEN alcohol IS NULL THEN 1 ELSE 0 END) AS alcohol_nulls,
SUM(CASE WHEN quality IS NULL THEN 1 ELSE 0 END) AS quality_nulls
FROM wine_cleaned;''', engine)
```

### Observation
- No nulls exist in all the features/columns.

## Final Conclusion
- All columns validated successfully
- No missing or invalid values
- Class imbalance is very high.A reliable metric should be choosen.
- Dataset is reliable for modeling
