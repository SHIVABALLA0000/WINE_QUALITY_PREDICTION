# 03. Univariate Feature Analysis

This notebook performs **univariate analysis for each physicochemical feature**.
- How wine type influences the individual physiocochemical features.
- Distributions of each feature.


## 1. Load Clean Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
from sqlalchemy import create_engine
engine= sqlite3.connect("wine_quality.db")
df = pd.read_sql('SELECT * FROM wine_cleaned', engine)

df.shape
```

## 2. Target Variable: Wine Quality

```python
plt.figure()
sns.countplot(x='quality', data=df)
plt.title('Wine Quality Distribution')
plt.show()
```

### Observation
- Wine quality is ordinal and imbalanced
- Middle quality classes dominate

## 3. Continuous Feature Distributions

```python
features = ['fixed_acidity','volatile_acidity','citric_acid','residual_sugar','chlorides','free_sulfur_dioxide','total_sulfur_dioxide','density','pH','sulphates','alcohol']

for col in features:
    plt.figure()
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

## 4. Feature Distributions by Wine Type (Interpretability Only)

```python
for col in features:
    plt.figure()
    sns.kdeplot(data=df, x=col, hue='wine_type', common_norm=False)
    plt.title(f'{col} Distribution by Wine Type')
    plt.show()
```

## 5. Summary of Univariate Insights

- Feature distributions are non-Gaussian
- Several features exhibit skewness and outliers
- Wine type explains distributional shifts but does not require dataset separation
- The combined dataset is suitable for modeling, with `wine_type` acting as an informative     categorical feature.
