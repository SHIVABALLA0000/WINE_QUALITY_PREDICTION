# 04. Bivariate Relationship Analysis 

This notebook performs ** bivariate analysis** using multiple plot types.

**Objective:** Robustly analyze feature–feature and feature–target relationships across all features.

## 1. Load Clean Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
import sqlite3

engine= sqlite3.connect("wine_quality.db")
df = pd.read_sql('SELECT * FROM wine_cleaned', engine)

df.shape
```

## 2. Feature × Feature Relationships

### Residual Sugar vs Density

```python
plt.figure()
sns.scatterplot(data=df, x='residual sugar', y='density', alpha=0.3)
plt.title('Residual Sugar vs Density (Scatter)')
plt.show()

plt.figure()
plt.hexbin(df['residual sugar'], df['density'], gridsize=40)
plt.colorbar(label='Count')
plt.title('Residual Sugar vs Density (Hexbin)')
plt.show()
```

- Scatter plot shows a curved, non-linear relationship:Density increases with residual sugar, but not linearly
- Hexbin reveals high-density regions at:low residual sugar + moderate density and clustering increases as sugar increases

### Alcohol vs Density

```python
plt.figure()
sns.scatterplot(data=df, x='alcohol', y='density', alpha=0.3)
plt.title('Alcohol vs Density')
plt.show()
```

- Clear inverse relationship indicates Higher alcohol content generally lowers density

### Free SO₂ vs Total SO₂

```python
plt.figure()
sns.scatterplot(data=df, x='free sulfur dioxide', y='total sulfur dioxide', alpha=0.3)
plt.title('Free SO₂ vs Total SO₂')
plt.show()
```

- Very strong linear relationship indicates Indicates strong multicollinearity
- Free sulfur dioxide is a component of total sulfur dioxide.

## 3. Feature × Target (Quality)

### Alcohol vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='alcohol', data=df)
plt.title('Alcohol vs Quality')
plt.show()

plt.figure()
sns.violinplot(x='quality', y='alcohol', data=df)
plt.title('Alcohol vs Quality (Violin)')
plt.show()
```

- Alcohol is a strong positive driver of quality.(boxplot)
- But from violin plot clearly shows overlapping which tells alcohol alone cannot separate classes.

### Sulphates vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='sulphates', data=df)
plt.title('Sulphates vs Quality')
plt.show()
```

- Slight upward trend in median sulphates and Very high variance across quality levels.
- Weak positive association with respective to quality

### Chlorides vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='chlorides', data=df)
plt.title('Chlorides vs Quality')
plt.show()
```

- Clearly there is no relationship exists between them i.e quality vs chlorides.

### Fixed Acidity vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='fixed acidity', data=df)
plt.title('Fixed Acidity vs Quality')
plt.show()
```

### Volatile Acidity vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='volatile acidity', data=df)
plt.title('Volatile Acidity vs Quality')
plt.show()
```

- There we can clearly observe negative relationship.
- so,volatile acidity is a negative indicator but also data overlaps.

### pH vs Quality

```python
plt.figure()
sns.boxplot(x='quality', y='pH', data=df)
plt.title('pH vs Quality')
plt.show()
```

- There is no relationship between quality and ph .may be have relationship between balanced acidity.

## 4. Interaction Effects

```python
plt.figure()
sns.scatterplot(data=df, x='alcohol', y='density', hue='quality', alpha=0.4)
plt.title('Alcohol vs Density by Quality')
plt.show()
```

- The alcohol–density plot colored by quality reveals a strong inverse and non-linear relationship.
- Higher-quality wines tend to exhibit higher alcohol and lower density.
- However,overlap across quality classes indicates that no single threshold can separate quality levels.

## 5. Summary
- Non-linear relationships dominate
- Heavy overlap across quality classes
- Single features are weak alone
- Interaction-aware models are required
