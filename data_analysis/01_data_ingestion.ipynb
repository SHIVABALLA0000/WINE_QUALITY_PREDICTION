# 01. Data Preprocessing 

This notebook focuses **only on data ingestion, validation, cleaning, feature engineering, and  clean data into SQLite**.


## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
import sqlite3
```

## 2. Load Raw Wine Quality Data

```python
red_wine = pd.read_csv(r'C:\Users\shiva\Downloads\wine_quality_prediction\data\winequality-red.csv', sep=';')
white_wine = pd.read_csv(r'C:\Users\shiva\Downloads\wine_quality_prediction\data\winequality-white.csv', sep=';')
```

## 3. Initial Data Validation

Basic validation checks are performed to ensure data reliability before any transformation.

```python

print('Red wine shape:', red_wine.shape)
print('White wine shape:', white_wine.shape)
```

## 4. Feature Engineering

A categorical feature `wine_type` is introduced to distinguish between red and white wines. This feature is created **before database insertion** to avoid schema mutation later.

```python
red_wine['wine_type'] = 'red'
white_wine['wine_type'] = 'white'

wine_df = pd.concat([red_wine, white_wine], ignore_index=True)
print('Combined dataset shape:', wine_df.shape)
```

## 5.Removing Duplicates

```python
wine_df.drop_duplicates(inplace=True)
print("Duplicates removed succesfully")
```

## 6.  Cleaned Data into SQLite Database

The cleaned and feature-engineered dataset is stored in a SQLite database. This database acts as the **single source of truth** for SQL analysis and modeling.

```python
engine = create_engine('sqlite:///data/database/wine_quality.db', echo=False)
engine= sqlite3.connect("wine_quality.db")
wine_df.drop_duplicates(inplace=True)
wine_df.to_sql('wine_cleaned', engine, if_exists='replace', index=False)

print('Cleaned data successfully stored in SQLite database.')
```



- SQLite database: `wine_quality.db`
- Table: `wine_cleaned`
